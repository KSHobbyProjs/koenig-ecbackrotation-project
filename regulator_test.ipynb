# Apply Regulator to Compute $r^2$
---
### Note: 
The idea was to compute the resonance at $\phi=0$ from the unrotated Hamiltonian (analytically, this would be the divergent Gamow state), then regulate the tail to compute $\langle r^2\rangle$ directly. This can't work, however, because the unrotated Hamiltonian can't produce the original divergent Gamow state computationally. That's why we need CS to begin with.

So, the entire result here is useless. I only keep this notebook here for posterity.

```python
from dvr import *
```

```python
# define system parameters
n = 256                      # number of DVR grid points
phi = pi / 24                # rotation angle
real_res = 1.606 - 1j*0.047  # expected value of resonance

gauss_s = lambda r: 2.0 * exp(-((r-3.0)/1.5)**2) # potential
```

```python
L = 50.0

# exact rr with phi large enough to expose resonance
DVRtest = DVR(n, L, rotation_angle=phi, potential=gauss_s)
eigval, eigstate = DVRtest.closest_to_resonance(real_res)
rr = DVRtest.compute_r2(eigstate[:, 0], rotate_rr=True)
print(f"Energy of resonance at phi={phi:.3f} is E={eigval[0].real:.3f} + j{eigval[0].imag:.3f}. rr={rr.real:.3f} + j{rr.imag:.3f}.")
```

```python
# rr with phi = 0
DVRtest = DVR(n, L, rotation_angle=0.0, potential=gauss_s)
eigval, eigstate = DVRtest.closest_to_resonance(real_res)
rr = DVRtest.compute_r2(eigstate[:, 0], rotate_rr=False)
print(f"Energy of resonance at phi=0 is E={eigval[0].real:.3f} + j{eigval[0].imag:.3f}. rr = {rr.real:.3f} + j{rr.imag:.3f}.")
```

```python
# add regulator
DVRtest = DVR(n, L, rotation_angle=0.0, potential=gauss_s)
eigval, eigstate = DVRtest.closest_to_resonance(real_res)
rr = DVRtest.compute_r2(eigstate[:, 0], rotate_rr=False, regulator=True)
print(f"(regulator) Energy of resonance at phi={phi:.3f} is E={eigval[0].real:.3f} + j{eigval[0].imag:.3f}. rr={rr.real:.3f} + j{rr.imag:.3f}.")
```
